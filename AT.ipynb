Imports

```python
import numpy as np
import math
```

Declare Variables

```python
ra = input("Digite seu RA: ")
sum_ra = sum(int(digit) for digit in ra)
g = 9.81
h0 = 3 * sum_ra
v0 = float(input("Digite a velocidade inicial (m/s): "))
```

Solution

```python
a = 0.5 * -g
b = v0
c = h0
```

```python
delta = b**2 -4*a*c
```

```python
if delta < 0:
    print("Sem solução")
else:
    t1= (-b + math.sqrt(delta)) / 2*a
    t2= (-b - math.sqrt(delta)) / 2*a
    max_time = max(t1, t2)
```

Print result

```python
print(f"{max_time} segundos")
```
